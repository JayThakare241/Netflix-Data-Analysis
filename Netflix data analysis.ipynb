```python
# imports

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,5)
```

```python
# load data

df = pd.read_csv("netflix.csv")
df.head()
```

```python
# data info

df.info()
df.isnull().sum()
df.describe(include="all")
```

```python
# handle duplicates

df = df.drop_duplicates()
df.shape
```

```python
# handling missing values

df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Unknown")
df['country'] = df['country'].fillna("Unknown")
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])

df.isnull().sum()
```

```python
# EDA

sns.countplot(data=df, x="type", palette="Set2")
plt.title("Movies vs TV Shows")
plt.show()
```

```python
# top geners

df['genres'] = df['listed_in'].apply(lambda x: x.split(", "))
genres = sum(df['genres'], [])
genre_counts = pd.Series(genres).value_counts().head(10)

sns.barplot(x=genre_counts.values, y=genre_counts.index, palette="viridis")
plt.title("Top 10 Most Common Genres")
plt.show()
```

```python
# content added per year

df['year_added'] = df['date_added'].dt.year

sns.countplot(data=df, x="year_added", palette="coolwarm")
plt.title("Content Added Per Year")
plt.xticks(rotation=60)
plt.show()
```

```python
# top 10 directors

top_directors = df['director'].value_counts().head(10)

sns.barplot(x=top_directors.values, y=top_directors.index, palette="Blues_d")
plt.title("Top 10 Directors with Most Netflix Titles")
plt.show()
```

```python
# word cloud of movie titles

text = " ".join(df[df['type']=="Movie"]['title'])
wordcloud = WordCloud(width=800, height=400, background_color="black").generate(text)

plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Word Cloud of Movie Titles")
plt.show()
```

```python
# country with most content 

country_count = df['country'].value_counts().head(10)

sns.barplot(x=country_count.values, y=country_count.index)
plt.title("Top Countries by Netflix Content")
plt.show()
```
